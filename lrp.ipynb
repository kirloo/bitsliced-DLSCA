```python
import matplotlib.pyplot as plt

import json

import numpy as np
```

```python
MODELS_FOLDER = "E:/Master/Avhandling/Models/models"

IMPL = "fixslice"
TARGET = "key"
BYTE = 1
ARCH = "zhang"
TRACE_START = 0
TRACE_END = 1000
SEED = 777

MODEL_NAME = f"{IMPL}-{TARGET}-byte{BYTE}-{ARCH}-{TRACE_START}_{TRACE_END}-s{SEED}"
```

```python
def load_snr(byte, pt):
    with open(f"misc/snr/fixslice/sbox_byte{byte}_pt{pt}.json", 'r') as f:
        data = json.load(f)
    return data['snr'][TRACE_START:TRACE_END]
```

```python
with open(f"lrp/{MODEL_NAME}.json", 'r') as f:
    attr_info = json.load(f)

with open("misc/samples/raw_trace.json", 'r') as f:
    trace = json.load(f)['trace']

x_values = np.arange(len(trace))[TRACE_START:TRACE_END]

fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(12.0, 6.0))

axs[0].plot(x_values,attr_info["avg_attr"], 'magenta')
axs[1].plot(x_values,trace[TRACE_START:TRACE_END])

axs[0].set_title("LRP")
axs[1].set_title("Sample trace")

plt.tight_layout()

plt.show()
```
