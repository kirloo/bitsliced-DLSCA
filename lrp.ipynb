```python
import matplotlib.pyplot as plt

import json

import numpy as np
```

```python
MODELS_FOLDER = "E:/Master/Avhandling/Models/models"

IMPL = "fixslice"
TARGET = "sbox"
BYTE = 1
ARCH = "zhang"
TRACE_START = 400
TRACE_END = 1500
SEED = 777

MODEL_NAME = f"{IMPL}-{TARGET}-byte{BYTE}-{ARCH}-{TRACE_START}_{TRACE_END}-s{SEED}"
```

```python
def load_snr(byte, pt):
    with open(f"snr/sbox_byte{byte}_pt{pt}.json", 'r') as f:
        data = json.load(f)
    return data['snr'][TRACE_START:TRACE_END]
```

```python
with open(f"lrp/{MODEL_NAME}.json", 'r') as f:
    attr_info = json.load(f)

with open("samples/raw_trace.json", 'r') as f:
    trace = json.load(f)['trace']

x_values = np.arange(len(trace))[TRACE_START:TRACE_END]

fig, axs = plt.subplots(nrows=3, ncols=1)
fig.set_figheight(8.0)
fig.set_figwidth(12.0)

axs[0].plot(x_values,attr_info["avg_abs_attr"], 'magenta')
axs[1].plot(x_values,load_snr(1,1), 'purple')
axs[2].plot(x_values, trace[TRACE_START:TRACE_END])

axs[0].set_title("Absolute LRP")
axs[1].set_title("SNR")
axs[2].set_title("Example Trace (raw)")

fig.set_tight_layout(True)

plt.show()
```
